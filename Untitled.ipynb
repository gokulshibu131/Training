```python
n=int(input("Enter the no. of elements:"))
arr=[]
for i in range(n):
  x=int(input(f"Enter element {i+1}: ")) # Prompt for one element at a time
  arr.append(x)

print("Original array:",arr)
print("Reversed array:",arr[::-1])
```

```python
arr=[]
n=int(input("enter array limit:"))
for i in range(n):
  arr.append(int(input("Enter a no:")))
  print("Sum of elements:",sum(arr))
```
